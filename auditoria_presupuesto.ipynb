```python
import pandas as pd

estados = pd.read_csv('AN_DES_resultados_estados.csv')

# 2. Creamos la columna de Variación Porcentual Automatizada
# Fórmula: ((Pagado - Aprobado) / Aprobado) * 100
estados['Variacion_Porcentual'] = (((estados['Total_Pagado_Pesos'] - estados['Total_Aprobado_Pesos']) / estados['Total_Aprobado_Pesos']) * 100)

# 3. Redondeamos a 2 decimales para que se vea limpio
estados = estados.round(2)

# 4. Ordenamos para ver a qué estados les aumentaron más el dinero (positivo)
# y a cuáles les recortaron más (negativo)
estados.sort_values(by='Variacion_Porcentual', ascending=False)
```
